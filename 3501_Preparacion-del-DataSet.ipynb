# Imports 

```python
import arff
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

## Funciones Auxiliares

```python
def load_kdd_dataset(data_path):
    """Lectura del conjunto de datos NSL-KDD."""
    with open(data_path, 'r') as train_set:
        dataset = arff.load(train_set)
    attributes = [attr[0] for attr in dataset["attributes"]]
    return pd.DataFrame(dataset["data"], columns=attributes)
```

```python
def train_val_test_split(df, rstate=42, shuffle=True, stratify=None):
    strat = df[stratify] if stratify else None
    train_set, test_set = train_test_split(
        df, test_size=0.4, random_state=rstate, shuffle=shuffle, stratify=strat)
    strat = test_set[stratify] if stratify else None
    val_set, test_set = train_test_split(
        test_set, test_size=0.5, random_state=rstate, shuffle=shuffle, stratify=strat)
    return (train_set, val_set, test_set)
```

# 1.- Lectura del Data Set

```python
df = load_kdd_dataset("datasets/NSL-KDD/KDDTrain+.arff")
```

#### 3.1 Imputer Sklearn 
Existe otra alternativa para la opcion 3, que consiste en usar la clase Imputer de Sklearn

```python
# copiar el Data set para no alterar el original el = es de asignacion
X_train_copy = X_train.copy()
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy = "median")
```

```python
# La clase imputer no admite valores categoricos, se elimina los atributos categoricos (los que no sean numericos
X_train_copy_num = X_train_copy.select_dtypes(exclude = ["object"])
X_train_copy_num.info()
```

```python
# proporcionar atributos numericos para que calcule los valores 
imputer.fit(X_train_copy_num)
```

```python
# rellenar los valores nulos 
X_train_copy_num_nonan = imputer.transform(X_train_copy_num)
```

```python
# transformar el resultado a un dataframe de pandas 
X_train_copy = pd.DataFrame(X_train_copy_num_nonan, columns = X_train_copy_num.columns)
X_train_copy.head(10)
```
