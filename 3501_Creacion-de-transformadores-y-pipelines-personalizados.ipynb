```python
import arff
import pandas as pd
import numpy as np 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler
```

## Funciones auxiliares 

```python
def load_kdd_dataset(data_path):
    """Lectura del conjunto de datos NSL-KDD."""
    with open(data_path, 'r') as train_set:
        dataset = arff.load(train_set)
    attributes = [attr[0] for attr in dataset["attributes"]]
    return pd.DataFrame(dataset["data"], columns=attributes)
```

```python
def train_val_test_split(df, rstate=42, shuffle=True, stratify=None):
    strat = df[stratify] if stratify else None
    train_set, test_set = train_test_split(
        df, test_size=0.4, random_state=rstate, shuffle=shuffle, stratify=strat)
    strat = test_set[stratify] if stratify else None
    val_set, test_set = train_test_split(
        test_set, test_size=0.5, random_state=rstate, shuffle=shuffle, stratify=strat)
    return (train_set, val_set, test_set)
```

## API Sklearn 

Antes de continuar vamos hacer una reseña sobre como funcionan las API´s de Sklearn:

* **Esimators**: Cualquier objeto que pueda estimar cualquier parametro: 
    * El propio estimador se forma mediante el metodo fit(), que siempre toma un dataset como argumento.
    * Cualquier otro parametro de este metodo, es un hiperparametro
* **Transformers**: son estimadores capaces de transformar el dataset (como inputer).
    * La trandformacion se realiza mediante el metodo transform
    * Reciben un DataSet como parametro de entrada.
* **Predictors**: son estimadores capaces de realizar predicciones.
    * La prediccion se realiza mediante el metodo predict().
    * Recibe un data set como entrada
    * Retorna un data set con las predicciones
    * Tiene un metodo score()  para evaluar el resultado de la prediccion.

```python
# Transformador diseñado para escalar de manera sencilla unicamente algunas columnas seleccinadas 
class CustomScaler(BaseEstimator, TransformerMixin):
    def __init__(self, attributes):
        self.attributes = attributes
    def  fit(self, X, y = None):
        return self # nothing else to do 
    def transform(self, X, y = None):
        X_copy = X.copy()
        scale_attr = X_copy[self.attributes]
        robust_scaler = RobustScaler()
        X_scaled = robust_scaler.fit_transform(scale_attr)
        X_scaled = pd.DataFrame(X_scaled, columns = self.attributes, index=X_copy.index)
        for attr in self.attributes:
            X_copy[attr] = X_scaled[attr]
        return X_copy


# el robust_scaler es para normalizar datos categoricos atipicos y los convierte en dataframe
```

## 6.- Construyendo pipelines personalizados

Los Pipelines permiten agrupar en un flujo de ejecucion todas las operaciones de transformacion que se necesita realizar sobre el dataset, esto facilita muchisisisisimo las transformaciones para diferentes datasets. Los aspectos a tener en cuenta de estas estructuras son:

* Recibe un dataset de pares (nombre, estimador)
* Todos, menos el ultimo deben ser transformadores
* El pipeline expone los mismos metodos que el mismo que el **último estimador**.
* Cuando se llama el metodo fit() del pipeline se llama secuencialmente al metodo fit_transform() de los estimadores y se les pasa de manera secuencial el output del siguiente. El ultimo invoca el método fit().


```python
# Construccion de un pipeline para los atributos numericos
from sklearn.pipeline import Pipeline 
from sklearn.preprocessing import RobustScaler 
from sklearn.impute import SimpleImputer 

num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("rbst_scaler", RobustScaler()),
])
```

```python
X_train_num = X_train.select_dtypes(exclude=["object"])
X_train_prep = num_pipeline.fit_transform(X_train_num)
X_train_prep = pd.DataFrame(X_train_prep, columns=X_train_num.columns, index= X_train_num.index)
```

A continuacion se presenta el metodo **ColumnTransformer**, que ejecuta todos los pipelines y concatena el resultado para ello, el resultado de los pipelines debe ser un valor numerico.


```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

num_attribs = list(X_train.select_dtypes(exclude= ["object"]))
cat_attribs = list(X_train.select_dtypes(include= ["object"]))

full_pipeline = ColumnTransformer([
    ("num", num_pipeline, num_attribs),
    ("cat", OneHotEncoder(), cat_attribs),
])
```

```python
X_train_prep = full_pipeline.fit_transform(X_train)
```

```python
X_train_prep = pd.DataFrame(X_train_prep, columns=list(pd.get_dummies(X_train)), index= X_train.index)
```
